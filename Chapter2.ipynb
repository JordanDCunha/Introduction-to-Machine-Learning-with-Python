## What Is Supervised Learning?

- Used when:

- Learn a model from the training data

- Once trained, the model:
  - Automates predictions

## 📘 2.1 Classification and Regression

There are two main types of supervised machine learning problems:
- **Classification**
- **Regression**

---

### 🔹 Classification

- Goal: predict a **class label**
- The label is chosen from a **predefined set of categories**

#### Examples
- Iris species classification:
  - setosa
  - versicolor
  - virginica
- Language detection for a website
- Email spam detection

---

### 🔹 Types of Classification

#### Binary Classification
- Exactly **two classes**
- Often framed as a **yes / no question**
- Example:
  - Spam vs not spam
  - Question: *“Is this email spam?”*

**Terminology**
- One class is called the **positive class**
- The other is the **negative class**
- “Positive” does **not** mean good or desirable
  - It simply refers to the class of interest
  - Example: in spam detection, *spam* is often the positive class
- Which class is positive is **domain-dependent**

---

#### Multiclass Classification
- More than **two classes**
- Example:
  - Iris dataset (3 species)
  - Language prediction (many possible languages)

---

### 🔹 Regression

- Goal: predict a **continuous value**
- Output is a:
  - Floating-point number
  - Real-valued quantity

#### Examples
- Predicting annual income based on:
  - Education
  - Age
  - Location
- Predicting crop yield based on:
  - Weather
  - Past yields
  - Workforce size

---

### 🔹 Key Difference: Continuity of Output

A simple way to tell classification and regression apart:

- **Regression**
  - Output is **continuous**
  - Small differences in prediction usually don’t matter much
  - Example:
    - $40,000 vs $40,001 income → almost the same

- **Classification**
  - Output is **discrete**
  - No concept of “closeness” between classes
  - Example:
    - A website is either in English or French
    - There is no language *between* English and French


## 📘 2.2 Generalization, Overfitting, and Underfitting

### 🔹 Generalization

- Goal of supervised learning:
  - Learn from **training data**
  - Make accurate predictions on **new, unseen data**
- When a model performs well on unseen data, it is said to **generalize**
- The ultimate objective is **high generalization performance**, not just high training accuracy

---

### 🔹 Training vs Test Performance

- Models are usually built to perform well on the **training set**
- If training and test data are similar:
  - Good training performance *may* imply good test performance
- However:
  - Very **complex models** can achieve extremely high (even perfect) training accuracy
  - This does **not** guarantee good performance on new data

---

### 🔹 Example: Predicting Boat Buyers

- Task:
  - Predict whether a customer will buy a boat
  - Use historical customer data
  - Send promotions only to likely buyers

- A novice data scientist creates the rule:
  > “If the customer is older than 45, and has less than 3 children or is not divorced, then they will buy a boat.”

- Result:
  - **100% accuracy** on the training data
  - Rule perfectly fits the existing dataset

---

### 🔹 Why Perfect Training Accuracy Is a Problem

- Many rules can perfectly explain the training data:
  - No age appears twice → rules could be based on exact ages
- But:
  - We already know the outcomes for training data
  - The real goal is predicting **new customers**
- The rule:
  - Is **too complex**
  - Is based on **very little data**
  - Some conditions rely on a single example

---

### 🔹 Overfitting

- **Overfitting** occurs when:
  - A model fits the training data **too closely**
  - It captures noise or peculiarities instead of true patterns
- Characteristics:
  - Excellent training performance
  - Poor performance on new data
- The boat-buying rule is an example of **overfitting**

---

### 🔹 Underfitting

- **Underfitting** happens when:
  - The model is **too simple**
  - It fails to capture important patterns in the data
- Example:
  - “Everyone who owns a house buys a boat”
- Characteristics:
  - Poor performance even on the training set

---

### 🔹 Model Complexity Trade-Off

- As model complexity increases:
  - Training accuracy improves
- If complexity becomes too high:
  - Generalization performance decreases
- There exists a **sweet spot**:
  - Model is complex enough to capture patterns
  - Simple enough to generalize well

- This balance is known as the:
  - **Overfitting–Underfitting Trade-off**

---

## 📘 2.2.1 Relation of Model Complexity to Dataset Size

### 🔹 Dataset Size Matters

- Model complexity depends on:
  - The **amount**
  - The **variety**
  of training data
- Larger and more diverse datasets:
  - Allow more complex models without overfitting
- Simply duplicating data:
  - Does **not** increase useful information

---

### 🔹 Revisiting the Boat Example

- If we had **10,000 additional customer records**
- And they all followed the same rule:
  - We would trust the complex rule much more
- More data → stronger evidence → better generalization

---

### 🔹 Key Takeaway

- More data often enables:
  - Better models
  - Better generalization
- In practice:
  - Collecting more data can be more effective than tuning models
- **Never underestimate the power of more data**

## 2.3.2 k-Nearest Neighbors (k-NN)

- k-NN is one of the simplest machine learning algorithms.
- Training consists of storing the dataset.
- Predictions are made by finding the closest training points.


```python
# One-nearest-neighbor classification visualization
mglearn.plots.plot_knn_classification(n_neighbors=1)
```

- With k = 1:
  - Prediction equals the label of the nearest point
- With larger k:
  - Majority voting determines the predicted class


```python
# Three-nearest-neighbors visualization
mglearn.plots.plot_knn_classification(n_neighbors=3)
```

### k-NN Classification with scikit-learn

- We split data into training and test sets.
- We evaluate generalization performance.


```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

X, y = mglearn.datasets.make_forge()
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

clf = KNeighborsClassifier(n_neighbors=3)
clf.fit(X_train, y_train)

print("Test set predictions:", clf.predict(X_test))
print("Test set accuracy: {:.2f}".format(clf.score(X_test, y_test)))
```

- Increasing k:
  - Simplifies the model
  - Smooths decision boundaries
- Small k → high complexity
- Large k → underfitting


```python
# Decision boundaries for different k values
fig, axes = plt.subplots(1, 3, figsize=(10, 3))

for n_neighbors, ax in zip([1, 3, 9], axes):
    clf = KNeighborsClassifier(n_neighbors=n_neighbors).fit(X, y)
    mglearn.plots.plot_2d_separator(clf, X, fill=True, eps=0.5,
                                    ax=ax, alpha=.4)
    mglearn.discrete_scatter(X[:, 0], X[:, 1], y, ax=ax)
    ax.set_title(f"{n_neighbors} neighbor(s)")
    ax.set_xlabel("feature 0")
    ax.set_ylabel("feature 1")

axes[0].legend(loc=3)
```

### k-NN Regression

- Regression version predicts:
  - Mean target value of nearest neighbors
- More neighbors → smoother predictions


```python
from sklearn.neighbors import KNeighborsRegressor

X, y = mglearn.datasets.make_wave(n_samples=40)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

reg = KNeighborsRegressor(n_neighbors=3)
reg.fit(X_train, y_train)

print("Test set predictions:\n", reg.predict(X_test))
print("Test set R^2: {:.2f}".format(reg.score(X_test, y_test)))
```

### Strengths and Weaknesses of k-NN

**Strengths**
- Simple and intuitive
- No training cost

**Weaknesses**
- Slow predictions for large datasets
- Performs poorly with many features or sparse data
- Sensitive to feature scaling


## 2.3.3 Linear Models

- Linear models predict using a weighted sum of features.
- Used for both regression and classification.
- Model complexity is controlled via regularization.


```python
# Linear regression on wave dataset
from sklearn.linear_model import LinearRegression

X, y = mglearn.datasets.make_wave(n_samples=60)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

lr = LinearRegression().fit(X_train, y_train)

print("lr.coef_:", lr.coef_)
print("lr.intercept_:", lr.intercept_)
print("Training set score: {:.2f}".format(lr.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lr.score(X_test, y_test)))
```

- Linear regression:
  - No regularization
  - Can overfit in high-dimensional settings


```python
# Ridge regression (L2 regularization)
from sklearn.linear_model import Ridge

ridge = Ridge().fit(X_train, y_train)
print("Training set score: {:.2f}".format(ridge.score(X_train, y_train)))
print("Test set score: {:.2f}".format(ridge.score(X_test, y_test)))
```

- Ridge regression:
  - Uses L2 regularization
  - Shrinks coefficients toward zero
  - Improves generalization


```python
# Lasso regression (L1 regularization)
from sklearn.linear_model import Lasso

lasso = Lasso(alpha=0.01, max_iter=100000).fit(X_train, y_train)
print("Training set score: {:.2f}".format(lasso.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lasso.score(X_test, y_test)))
print("Number of features used:", np.sum(lasso.coef_ != 0))
```

- Lasso:
  - Uses L1 regularization
  - Sets many coefficients exactly to zero
  - Performs automatic feature selection


### Linear Models for Classification

- Common linear classifiers:
  - Logistic Regression
  - Linear Support Vector Machines (LinearSVC)
- Use regularization parameter C:
  - Large C → complex model
  - Small C → simpler model


```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

X, y = mglearn.datasets.make_forge()

fig, axes = plt.subplots(1, 2, figsize=(10, 3))

for model, ax in zip([LinearSVC(), LogisticRegression()], axes):
    clf = model.fit(X, y)
    mglearn.plots.plot_2d_separator(clf, X, fill=False,
                                    eps=0.5, ax=ax)
    mglearn.discrete_scatter(X[:, 0], X[:, 1], y, ax=ax)
    ax.set_title(clf.__class__.__name__)
```

### Key Takeaways

- Model complexity must be balanced to avoid overfitting and underfitting.
- Regularization is essential for linear models.
- k-NN is simple but rarely used in production.
- Linear models scale well and are widely used in practice.


## 2.3.4 Naive Bayes Classifiers

Naive Bayes classifiers are a family of fast and simple supervised learning models.
They are similar to linear models but train even faster, at the cost of slightly worse
generalization performance.

They work by modeling each feature independently and collecting simple per-class
statistics.

### Types of Naive Bayes in scikit-learn
- **GaussianNB**: for continuous features
- **BernoulliNB**: for binary features
- **MultinomialNB**: for count-based features (e.g., word counts)

BernoulliNB and MultinomialNB are commonly used for text classification.


```python
import numpy as np

X = np.array([[0, 1, 0, 1],
              [1, 0, 1, 1],
              [0, 0, 0, 1],
              [1, 0, 1, 0]])
y = np.array([0, 1, 0, 1])
```

For BernoulliNB, we count how often each feature is nonzero for each class.


```python
counts = {}
for label in np.unique(y):
    counts[label] = X[y == label].sum(axis=0)

print("Feature counts:\n", counts)
```

### Strengths and Weaknesses of Naive Bayes

**Strengths**
- Extremely fast training and prediction
- Works well with high-dimensional sparse data
- Robust to parameter choices

**Weaknesses**
- Often slightly less accurate than linear models
- Assumes feature independence (often violated in practice)

Naive Bayes models are excellent baseline models, especially for large datasets.


## 2.3.5 Decision Trees

Decision trees are models for classification and regression that learn a sequence
of if/else rules.

Each node asks a question about a feature, and each leaf makes a prediction.


### How Decision Trees Work

- Each split tests a condition like:  
  **"Is feature i ≤ value a?"**
- The algorithm chooses the most informative split
- The process is repeated recursively
- Leaves become *pure* when they contain only one class


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data, cancer.target, stratify=cancer.target, random_state=42)

tree = DecisionTreeClassifier(random_state=0)
tree.fit(X_train, y_train)

print("Training accuracy:", tree.score(X_train, y_train))
print("Test accuracy:", tree.score(X_test, y_test))
```

Unrestricted trees often overfit.
To reduce overfitting, we use **pre-pruning**, such as limiting tree depth.


```python
tree = DecisionTreeClassifier(max_depth=4, random_state=0)
tree.fit(X_train, y_train)

print("Training accuracy:", tree.score(X_train, y_train))
print("Test accuracy:", tree.score(X_test, y_test))
```

### Feature Importance in Trees

Decision trees provide `feature_importances_`, which indicate how useful each
feature is for prediction.

- Values range from 0 to 1
- All importances sum to 1


```python
tree.feature_importances_
```

## 2.3.6 Ensembles of Decision Trees – Random Forests

Random forests reduce overfitting by averaging many decision trees.

Each tree:
- Uses a bootstrap sample of the data
- Considers only a subset of features at each split


```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(n_estimators=100, random_state=0)
forest.fit(X_train, y_train)

print("Training accuracy:", forest.score(X_train, y_train))
print("Test accuracy:", forest.score(X_test, y_test))
```

### Strengths of Random Forests

- Excellent performance with little tuning
- Resistant to overfitting
- No feature scaling required

**Key parameters**
- `n_estimators`
- `max_features`
- `max_depth`


## Gradient Boosting

Gradient boosting builds trees sequentially.
Each tree corrects the mistakes of the previous ones.

- Uses shallow trees
- Requires careful tuning
- Often achieves very high accuracy


```python
from sklearn.ensemble import GradientBoostingClassifier

gbrt = GradientBoostingClassifier(random_state=0)
gbrt.fit(X_train, y_train)

print("Training accuracy:", gbrt.score(X_train, y_train))
print("Test accuracy:", gbrt.score(X_test, y_test))
```

## 2.3.7 Kernelized Support Vector Machines (SVMs)

Kernelized SVMs allow nonlinear decision boundaries using the **kernel trick**.

The most common kernel is the **RBF (Gaussian) kernel**.


```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(X_train, y_train)

print("Training accuracy:", svc.score(X_train, y_train))
print("Test accuracy:", svc.score(X_test, y_test))
```

SVMs are very sensitive to feature scales.
We must scale features before training.


```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svc = SVC(C=1000)
svc.fit(X_train_scaled, y_train)

print("Training accuracy:", svc.score(X_train_scaled, y_train))
print("Test accuracy:", svc.score(X_test_scaled, y_test))
```

### Summary

- **Random Forests**: Robust, high accuracy, little preprocessing

### 2.4 Uncertainty Estimates from Classifiers

- In many applications, it’s important not only to predict a class, but also to know **how confident** the model is.
- Different prediction errors can have very different consequences (e.g., false positives vs. false negatives in medical diagnosis).
- scikit-learn provides two main methods for uncertainty estimation:
  - `decision_function`
  - `predict_proba`
- Most classifiers implement at least one of these methods, and many implement both.


### Example Setup: Gradient Boosting on a Synthetic Dataset

- We use a 2D synthetic dataset generated with `make_circles`.
- Classes are renamed for clarity (`blue` and `red`).
- A `GradientBoostingClassifier` is trained, which supports both:
  - `decision_function`
  - `predict_proba`


```python
import numpy as np
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split

X, y = make_circles(noise=0.25, factor=0.5, random_state=1)

# rename classes for illustration
y_named = np.array(["blue", "red"])[y]

X_train, X_test, y_train_named, y_test_named, y_train, y_test = \
    train_test_split(X, y_named, y, random_state=0)

gbrt = GradientBoostingClassifier(random_state=0)
gbrt.fit(X_train, y_train_named)
```

### 2.4.1 The Decision Function (Binary Classification)

- `decision_function` returns a **single score per sample** in binary classification.
- Shape: `(n_samples,)`
- Interpretation:
  - Positive values → preference for the positive class
  - Negative values → preference for the negative class
- The magnitude reflects how strongly the model prefers a class.


```python
print("X_test.shape:", X_test.shape)
print("Decision function shape:", gbrt.decision_function(X_test).shape)

# show first few decision scores
print("Decision function:", gbrt.decision_function(X_test)[:6])
```

- Predictions can be recovered by checking whether the decision score is greater than 0.
- The predicted class depends on the `classes_` attribute:
  - `classes_[0]` → negative class
  - `classes_[1]` → positive class


```python
print("Thresholded decision function:\n",
      gbrt.decision_function(X_test) > 0)

print("Predictions:\n", gbrt.predict(X_test))
```

- To fully reconstruct predictions:
  - Convert boolean output to integers (0 or 1)
  - Use them as indices into `classes_`


```python
greater_zero = (gbrt.decision_function(X_test) > 0).astype(int)
pred = gbrt.classes_[greater_zero]

print("pred is equal to predictions:",
      np.all(pred == gbrt.predict(X_test)))
```

- The scale of `decision_function` values is **arbitrary**.
- It depends on:
  - The model
  - The dataset
  - Hyperparameters
- This makes decision scores harder to interpret directly.


```python
decision_function = gbrt.decision_function(X_test)

print(
    "Decision function minimum: {:.2f} maximum: {:.2f}".format(
        decision_function.min(),
        decision_function.max()
    )
)
```

### 2.4.2 Predicting Probabilities

- `predict_proba` returns **class probabilities** instead of arbitrary scores.
- Shape for binary classification:
  - `(n_samples, 2)`
- Properties:
  - Values are between 0 and 1
  - Probabilities for each sample sum to 1
- Often easier to interpret than `decision_function`.


```python
print("Shape of probabilities:", gbrt.predict_proba(X_test).shape)

print("Predicted probabilities:")
print(gbrt.predict_proba(X_test[:6]))
```

- The predicted class is the one with probability > 0.5.
- Model certainty depends on:
  - Model complexity
  - Degree of overfitting
- A well-calibrated model:
  - Predictions made with 70% confidence are correct ~70% of the time.


```python
### 2.4.3 Uncertainty in Multiclass Classification

- Both `decision_function` and `predict_proba` also work for multiclass problems.
- Output shape:
  - `(n_samples, n_classes)`
- Each column corresponds to one class.
- Predictions are obtained using `argmax` across classes.
```

```python
from sklearn.datasets import load_iris

iris = load_iris()

X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, random_state=42
)

gbrt = GradientBoostingClassifier(learning_rate=0.01, random_state=0)
gbrt.fit(X_train, y_train)
```

```python
print("Decision function shape:", gbrt.decision_function(X_test).shape)
print("Decision function:")
print(gbrt.decision_function(X_test)[:6])
```

- For multiclass problems:
  - The predicted class is the one with the **maximum score**.
- This applies to both:
  - `decision_function`
  - `predict_proba`


print("Argmax of decision function:")
print(np.argmax(gbrt.decision_function(X_test), axis=1))

print("Predictions:")
print(gbrt.predict(X_test))


- When class labels are strings or non-consecutive integers:
  - Always use the `classes_` attribute
- This ensures predictions align with actual class names.


```python
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(max_iter=1000)

named_target = iris.target_names[y_train]
logreg.fit(X_train, named_target)

argmax_dec_func = np.argmax(logreg.decision_function(X_test), axis=1)

print("Classes:", logreg.classes_)
print("Predictions:", logreg.predict(X_test)[:10])
print("Recovered predictions:",
      logreg.classes_[argmax_dec_func][:10])
```
