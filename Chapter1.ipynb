# Chapter 1. Introduction (Summary)

Machine learning is the study of extracting useful knowledge from data. It combines ideas from statistics, artificial intelligence, and computer science, and is also known as predictive analytics or statistical learning.

Machine learning is widely used in everyday life. Many popular platforms such as Facebook, Amazon, and Netflix rely on multiple machine learning models to power features like recommendations, personalization, and image recognition.

Beyond commercial use, machine learning plays a major role in modern scientific research. It has been applied to fields such as astronomy, physics, biology, and medicine, enabling discoveries like finding new planets, analyzing DNA, and improving cancer treatments.

This chapter explains why machine learning has become so important, what types of problems it can solve, and introduces the key ideas needed to build a first machine learning model.


## 1.1 Why Machine Learning?

- Early intelligent systems used hand-coded *if–else* rules.
- Rule-based systems work only for simple, well-understood tasks.
- Major limitations of hand-coded rules:
  - Logic is specific to a single task and hard to reuse.
  - Requires deep human expert knowledge.
- Some problems (e.g., face detection) are impossible to solve with rules.
- Computers represent data differently than humans.
- Machine learning learns patterns automatically from data instead of rules.

## 1.1.1 Problems Machine Learning Can Solve

### Supervised Learning
- Learns from labeled input–output pairs.
- Can make predictions on unseen data.
- Requires a “teacher” providing correct outputs.
- Performance is easy to measure.
- Data labeling can be expensive or time-consuming.
- Examples:
  - Handwritten zip code recognition
  - Tumor classification from medical images
  - Credit card fraud detection
- Data collection difficulty varies by domain:
  - Manual labeling can be easy but slow.
  - Medical data requires experts and expensive equipment.
  - Fraud data can be collected automatically from user reports.

### Unsupervised Learning
- Uses only input data (no labeled outputs).
- Finds patterns or structure in data.
- Harder to evaluate than supervised learning.
- Examples:
  - Topic discovery in text documents
  - Customer segmentation
  - Anomaly detection in website traffic

## Data Representation and Features
- Data is often represented as a table:
  - Rows = samples (data points)
  - Columns = features (properties)
- Features describe the information available to the model.
- Good feature selection is critical for accurate predictions.
- Models cannot predict information that is not present in the data.
- Feature engineering improves how data is represented.

## 1.1.2 Knowing Your Task and Knowing Your Data
- Understanding the problem is more important than choosing an algorithm.
- Important questions to consider:
  - What problem am I trying to solve?
  - Can my data answer this question?
  - Do I have enough data?
  - Are the chosen features meaningful?
  - How will success be measured?
  - How does the model fit into the larger system?
- Machine learning is only one part of solving a real-world problem.

# 1.7 A First Application: Classifying Iris Species

- Goal: build a machine learning model that predicts the **species of an iris flower**
- Available measurements (features):
  - Sepal length (cm)
  - Sepal width (cm)
  - Petal length (cm)
  - Petal width (cm)
- Possible species (classes):
  - setosa
  - versicolor
  - virginica
- This is a **supervised learning** problem because the correct species is known
- This is a **classification problem** because we predict a discrete label
- Since there are three species, this is a **three-class classification** task
- Each flower is a **sample**
- The species of a flower is its **label**


## 1.7.1 Meet the Data

- Dataset used: **Iris dataset**
- A classic dataset in machine learning and statistics
- Included directly in **scikit-learn**
- Loaded using the `load_iris` function
- The returned object is a **Bunch**, similar to a dictionary
- Important keys:
  - `data`: feature measurements
  - `target`: numeric labels
  - `target_names`: species names
  - `feature_names`: descriptions of features
  - `DESCR`: dataset description


## Dataset Structure

- Number of samples: **150 flowers**
- Number of features: **4**
- Data shape follows the convention:
  - `(number of samples, number of features)`
- For the Iris dataset:
  - `(150, 4)`
- Feature names:
  - Sepal length (cm)
  - Sepal width (cm)
  - Petal length (cm)
  - Petal width (cm)


## Target Labels

- The `target` array contains the species for each flower
- It is a **one-dimensional NumPy array**
- One label per sample
- Labels are encoded as integers:
  - `0` → setosa
  - `1` → versicolor
  - `2` → virginica
- The mapping is stored in `target_names`


```python
# 🧪 Load the Dataset
from sklearn.datasets import load_iris

iris_dataset = load_iris()
```

## 📘 Dataset Structure

- Number of samples: **150 flowers**
- Number of features: **4**

**Data shape format:**
(number_of_samples, number_of_features)

**Here:**
(150, 4)

**Feature names:**
- Sepal length (cm)
- Sepal width (cm)
- Petal length (cm)
- Petal width (cm)


```python
# 🧪 Inspect the Data
print("Feature names:\n", iris_dataset['feature_names'])
print("Shape of data:", iris_dataset['data'].shape)
print("First five rows of data:\n", iris_dataset['data'][:5])
```

## 📘 Target Labels

- `target` is a **1D NumPy array**
- One label per flower
- Encoded as integers:
  - 0 → setosa
  - 1 → versicolor
  - 2 → virginica
- Mapping stored in `target_names`


```python
# 🧪 Inspect Targets
print("Target names:", iris_dataset['target_names'])
print("Shape of target:", iris_dataset['target'].shape)
print("Target:\n", iris_dataset['target'])
```

## 1.7.2 Measuring Success: Training and Testing Data

- We cannot evaluate a model using the same data it was trained on
- A model could simply **memorize** the training data
- To evaluate performance, data is split into:
  - **Training set**: used to build the model
  - **Test set**: used to evaluate the model
- Common split:
  - 75% training data
  - 25% test data
- Naming convention in scikit-learn:
  - `X`: feature matrix (2D)
  - `y`: target vector (1D)
- Data is shuffled before splitting to ensure all classes appear in both sets
- `random_state` is fixed to make results reproducible


```python
# 🧪 Train/Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'],
    iris_dataset['target'],
    random_state=0
)
```

## Training and Test Set Sizes

- Training set:
  - 112 samples
  - 4 features per sample
- Test set:
  - 38 samples
  - 4 features per sample


# 🧪 Check Shapes
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)


## 1.7.3 First Things First: Look at Your Data

- Inspecting data before modeling is critical
- Helps identify:
  - Errors
  - Inconsistencies
  - Unexpected measurements
- Visualization is a powerful inspection tool
- A **scatter matrix (pair plot)** shows:
  - All pairs of features
  - Data points colored by class
- The diagonal contains histograms of each feature
- From the plots:
  - The three classes are well separated
  - A machine learning model should perform well


```python
# 🧪 Scatter Matrix
import pandas as pd
import mglearn

iris_dataframe = pd.DataFrame(
    X_train,
    columns=iris_dataset.feature_names
)

pd.plotting.scatter_matrix(
    iris_dataframe,
    c=y_train,
    figsize=(15, 15),
    marker='o',
    hist_kwds={'bins': 20},
    s=60,
    alpha=.8,
    cmap=mglearn.cm3
)
```

## 1.7.4 Building Your First Model: k-Nearest Neighbors

- Algorithm used: **k-Nearest Neighbors (k-NN)**
- Idea:
  - Find the closest training points to a new sample
  - Predict the majority class
- The parameter `k` controls how many neighbors are considered
- In this example:
  - `k = 1`
  - Only the single closest neighbor is used
- Implemented in scikit-learn as `KNeighborsClassifier`
- Training consists of storing the training data


```python
# 🧪 Build & Train Model
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)
```

## 1.7.5 Making Predictions

- New data must be provided as a **2D array**
- Even a single sample must have shape:
  - `(1, number of features)`
- The model predicts:
  - A numeric label
  - Which can be mapped to the species name


```python
# 🧪 Predict New Sample
import numpy as np

X_new = np.array([[5, 2.9, 1, 0.2]])
prediction = knn.predict(X_new)

print("Prediction:", prediction)
print(
    "Predicted target name:",
    iris_dataset['target_names'][prediction]
)
```

## 1.7.6 Evaluating the Model

- Evaluation is done using the **test set**
- Predictions are compared to true labels
- Performance metric used:
  - **Accuracy**
- Accuracy = fraction of correct predictions
- The k-NN model achieves:
  - **97% accuracy**
- This indicates the model generalizes well to unseen data


```python
# 🧪 Model Evaluation
y_pred = knn.predict(X_test)

print("Test set score: {:.2f}".format(np.mean(y_pred == y_test)))
print("Test set score: {:.2f}".format(knn.score(X_test, y_test)))
```

## Core scikit-learn Workflow

The following code shows the **essential pattern** for training and evaluating
a supervised machine learning model using scikit-learn.


```python
# 🧪 Summary Code: Training and Evaluation Pipeline

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    iris_dataset['data'],
    iris_dataset['target'],
    random_state=0
)

# Build and train model
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)

# Evaluate model
print("Test set score: {:.2f}".format(knn.score(X_test, y_test)))
```
